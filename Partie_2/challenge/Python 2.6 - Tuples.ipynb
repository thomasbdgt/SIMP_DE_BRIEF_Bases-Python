# Mission 4

Create 2 tuples of the same length that contain only integers.

Then create a script that compares the sum of their elements and displays the tuple that has the highest total.

```python
tuple1 = (1, 2, 3, 4, 5)
tuple2 = (6, 7, 8, 9, 10)
somme1 = 0
somme2 = 0

for i in tuple1:
  somme1 += i
for j in tuple2:
  somme2 += j

if i > j:
  print(f"tuple1 has the highest total with {somme1} \n Here it is : {tuple1}")
elif j > i:
  print(f"tuple2 has the highest total with {somme2} \n Here it is : {tuple2}")
else:
  print(f"Both tuples have the same total with {somme1} \n Here are the tuples : \n Tuple 1 : {tuple1} \n Tuple 2 : {tuple2}")
```

# Mission 5

Here's a tuple:

`my_tuple = ("data analyst", "data scientist", "data engineer", "data architect")`

Ask the user for a string, then for an integer position. Create an altered copy of `my_tuple` where the element at the given position is now the given string.

```python
my_tuple = ("data analyst", "data scientist", "data engineer", "data architect")
string = str(input("Please insert any string : "))
nb = int(input("Please choose a number between 0 and 3 (included) : "))

if nb >= len(my_tuple):  # Si le nb renseigné dépasse la taille du tuple
  print("Problème de longueur")

alteration = list(my_tuple)
alteration[nb] = string

my_tuple = tuple(alteration)

print(my_tuple)
```
