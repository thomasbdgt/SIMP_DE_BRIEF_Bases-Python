# Mission 1

Below is a dictionary containing the number of French campuses that offer each kind of Wild Code School course.

wildcodeschool = {
   "data_analyst": 4,
   "data_scientist": 2,
   "dev_web": "13"
}

Find a way to only display the keys that have int values.

```python
wildcodeschool =	{
  "data_analyst": 4,
  "data_scientist": 2,
  "dev_web": "13"
}
```

```python

for key in wildcodeschool:  # On utilise un boucle for qui va passer en revue les clés du dictionnaire
  if isinstance(wildcodeschool.get(key), int):  # On vérifie que le type des valeurs contenues dans le dictionnaire sont des integers
    print(key)  # Si c'est le cas on affiche les clés.
```

# Mission 2

For the same dictionary, find a way to display all the values that have a key that contains the letter "e".

```python
for key in wildcodeschool:  #  On parcourt les clés du dictionnaire
  key = str(key)  # On convertit les clés en str pour pouvoir vérifier si elles contiennent "e"
  if key.find("e") >= 0:  # Si il y a un "e" dans la clé on obtient l'index (soit un nb >= 0), sinon on obtient -1
    print(wildcodeschool.get(key))
```

# Mission 3

Within the same dictionary, change the type of the value of the "dev_web" key to be an int. Then add a new key-value pair which will be "total_campus" : 19, using a **dictionary method**.

**Change the value of "dev_web" to be and int :**

```python
wildcodeschool["dev_web"] = int(wildcodeschool["dev_web"])  # On convertit la valeur associée à la clé "dev_web" en int

if isinstance(wildcodeschool["dev_web"], int):  # On vérifie que le type a bien été changé grâce à isinstance
  print(f"Value type has been succesfully changed to : {type(wildcodeschool["dev_web"])}")
else:
  print("Couldn't change the value type")
```

**Add a new key-value pair which will be "total_campus" : 19, using a dictionary method**

```python
wildcodeschool.update({"total_campus": 19}) # On utilise la méthode .update() pour ajouter les éléments demandés au dictionnaire
# wildcodeschool["total_campus"] = 19

print(wildcodeschool) # On print par curiosité, pour vérifier que l'ajout à bien eu lieu
```

# Mission 4

Take the dictionary generated from mission 3. Delete the key value pair :
- 'dev_web': 13 using a method
- 'total_campus': 19 using another method

```python
newdict = wildcodeschool.copy() # Je crée une copie pour ne pas altérer le dictionnaire de base

newdict.pop("dev_web")
del newdict["total_campus"]

print(newdict)
```

# Mission 5

Still within the dictionary generated from Mission 3, find a way to display the dictionary keys in alphabetical order.

```python
# Méthode 1 : On utilise une liste pour trier les clés et les afficher, mais on ne change pas réellement l'ordre dans le dicctionnaire
newdict2 = wildcodeschool.copy()  # Je crée une copie pour ne pas altérer le dictionnaire de base
alph_order = [] # Je crée une liste vide dans laquelle j'ajouterai par la suite les clés du dictionnaire

for i in newdict2:  # Je passe en revue les clés du dico...
  alph_order.append(i)  # ... et les ajoute une à une à dans la liste prévue à cet effet

alph_order.sort() # Je trie la liste par ordre alphabétique
print(alph_order)
```

# Mission 6

Create a script that returns the same elements as the **values()** method, as a list.

Example: for `{'a': 1, 'b': 2, 'c': 3}` it will create `[1, 2, 3]`

```python
dico = {'a': 1, 'b': 2, 'c': 3}
list_value = []
for i in dico:  # Boucle qui passe en revue les clés du dico...
  list_value.append(dico.get(i))  # ... et ajoute les valeurs correspondantes à ma liste

print(f"Voici ma liste : {list_value}")
```
