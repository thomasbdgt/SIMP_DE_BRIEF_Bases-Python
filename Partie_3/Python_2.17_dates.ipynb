# Your mission is to determine which date it was / it will be

```python
# what month will it be in 7 weeks ?
now_time = datetime.now()
seven_weeks = now_time + timedelta(weeks = 7)
print(f"In seven weeks, we'll be in {datetime.strftime(seven_weeks, '%B')}")
```

```python
# what day of the week will it be in 3 days ?
three_days = now_time + timedelta(days = 3)
print(f"In three days, it will be {datetime.strftime(three_days, '%A')}")
```

```python
# what hour will it be in 100 hours ?
hundred_hours = now_time + timedelta(hours = 100)
print(f"In one hundred hours, it will be {datetime.strftime(hundred_hours, '%H')}")
```

```python
# what date was 30 days ago ?
thirty_days_ago = now_time - timedelta(days = 30)
print(f"Thirty days ago, it was {datetime.strftime(thirty_days_ago, '%d/%m/%Y')}")
```

```python
# what date was 30 weeks ago ?
thirty_weeks_ago = now_time - timedelta(weeks = 30)
print(f"30 days ago, it was {datetime.strftime(thirty_weeks_ago, '%d/%m/%Y')}")
```

# With Pandas
"DATE" column has only string values. On a chart, the X-axis is not well displayed.

```python
# Please run this cell without modification:
df = pd.read_csv(
    "https://raw.githubusercontent.com/murpi/wilddata/master/quests/weather2019.csv"
)
fig, ax = plt.subplots(figsize=(20, 10))
ax.scatter(x=df["DATE"], y=df["MAX_TEMPERATURE_C"])
fig.show()
```

```python
# Convert the "DATE" column into a datetime format, then display the same chart. Are ticks on X-axis better printed ?
df_copy = df.copy()
df_copy['DATE'] = df_copy['DATE'].apply(lambda x: datetime.strptime(x, '%Y-%m-%d'))
```

```python
# Display only rows for the month of June
fig, ax = plt.subplots(figsize=(20, 10))
ax.scatter(x=df_copy['DATE'],
           y=df_copy['MAX_TEMPERATURE_C'])
fig.show()
```
