- Which actor is mentioned the most often?


```python
import pandas as pd

link = 'https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/imdb_1000.csv'
pd.read_csv(link)
```

```python
df = pd.read_csv(link)
```

```python
def nettoyage(chaine):
  liste_nettoyage = ["[u'",
                     "']",
                     "', ",
                     '[u"',
                     '"]',
                     '", ']
  for i in liste_nettoyage:
    if i in chaine:
      chaine = chaine.replace(i, "")
  return chaine

df['actors_list'] = df["actors_list"].map(lambda x: nettoyage(x).split("u'"))

classement = df.explode("actors_list").groupby("actors_list").title.count().sort_values(ascending = False)

classement
```
