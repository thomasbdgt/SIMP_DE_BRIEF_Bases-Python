```python
link = 'https://raw.githubusercontent.com/chriszapp/datasets/main/books.csv'
books = pd.read_csv(link, nrows = 13)
```

```python
# Mission 1: use apply to create a new column called 'reviews_opinion': if the ratings is between 4.5 and 5, the reviews_opinion is 'very good';
# if the rating is between 4 and 4.5, the reviews_opinion is 'good'; if the rating is between 3 and 4, the reviews_opinions will be 'neutral';
# if the rating is up to 3 the reviews_opinion is going to be 'bad'

# Add your code here
def reviews_opinion_apply(row):
    if row['average_rating'] < 3:
        return 'bad'
    elif 3 <= row['average_rating'] < 4:
        return 'neutral'
    elif 4 <= row['average_rating'] < 4.5:
        return 'good'
    else:
        return 'very good'

books['reviews_opinion'] = books.apply(reviews_opinion_apply, axis = 1)
books.loc[0:5, ['average_rating', 'reviews_opinion']]
```

```python
# Mission 2: using apply with a function, create a column with the number of authors per book
# (note that in the column 'authors', each name of the author is divided with the character '/')

# Add your code here
def nb_authors(authors):
    return len(authors.split(sep = '/'))

books['number_authors'] = books.authors.apply(nb_authors)
books.loc[0:5, ['authors', 'number_authors']]
```

```python
# Mission 3: rewrite the same task as Mission 2 using the lamdba function

# Add your code here
books['number_authors'] = books.authors.apply(lambda x: len(x.split(sep = '/')))
books.loc[0:5, ['authors', 'number_authors']]
```

```python
# Mission 4: create a column 'comment_ration'.
# Use the columns 'ratings_count' and 'text_reviews_count' to calculate the percentage of how many text review per reviews are per book:
# percentage = text / reviews * 100
# Use the mathematical operations between columns

# Add your code here
books["reviews_percentage"] = round((books['text_reviews_count'] / books['ratings_count']) * 100, 2)

books[['ratings_count', 'text_reviews_count', 'reviews_percentage']]
```

```python
# Mission 5: obtain the same result of Mission 4 using apply with a function.
# Try both methods to apply to the whole dataset and apply to selected columns

# Add your code here
def text_per_reviews(row):
    return round((row['text_reviews_count'] / row['ratings_count']) * 100, 2)

books["reviews_percentage"] = books.apply(text_per_reviews, axis = 1)

books[['ratings_count', 'text_reviews_count', 'reviews_percentage']]
```

```python
def text_per_reviews(row):
    return round((row.iloc[1] / row.iloc[0]) * 100, 2)

books["reviews_percentage"] = books[['ratings_count', 'text_reviews_count']].apply(text_per_reviews, axis = 1)

books[['ratings_count', 'text_reviews_count', 'reviews_percentage']]
```

```python
# Mission 6: obtain the same result of Mission 5 using apply lambda
# Try both methods to apply to the whole dataset and apply to selected columns

# Add your code here
books["reviews_percentage"] = books.apply(
    lambda x: round((x['text_reviews_count'] / x['ratings_count']) * 100, 2),
    axis = 1)

books[['ratings_count', 'text_reviews_count', 'reviews_percentage']]
```

```python
books["reviews_percentage"] = books[['ratings_count', 'text_reviews_count']].apply(
    lambda row: round((row.iloc[1] / row.iloc[0]) * 100, 2),
    axis = 1)

books[['ratings_count', 'text_reviews_count', 'reviews_percentage']]
```

```python
# Mission 7: extract the year from the publication_date column. How many years of publications are recorded in this dataset? Which are them?
# Add your code here
books["publication_year"] = books.publication_date.map(lambda x: int(x[-4:]))
books.publication_year.sort_values().unique()
```

```python
# Méthode avec datetime :
books["publication_year_dt"] = books.publication_date.map(lambda x: datetime.strftime(datetime.strptime(x, "%m/%d/%Y"), "%Y"))
books.publication_year_dt.sort_values().unique()
```
