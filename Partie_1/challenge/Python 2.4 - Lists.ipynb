# Mission 1

Define a list, then display only the elements (and their position) that have even positions (0, 2, 4, 6...).

With `test_list = ["a", "b", "c", "d", "e"]`, the script should display:
```
a at position 0
c at position 2
e at position 4
```

````python
test_list = ["a", "b", "c", "d", "e"]

# Your code here
for position, element in enumerate(test_list):
  if position % 2 == 0:
    print(f"{element} at position {position}")
````

# Mission 3

Define a list with five numbers, then ask the user for a number. Put the number at the end of the list, remove the first element from the list, and finally display the list.

For example, with the list `test_list3 = [1, 9, 6, 15, 4]` and the input `7`, you should display:
```
[9, 6, 15, 4, 7]
```


````python
list2 = [1, 9, 6, 15, 4]

user_nb = int(input("Specify the number you want to add to the list"))  # On demande un nombre à ajouter à la liste

list2 = list2[1:] + [user_nb] # On remplace liste en omettant la première valeur (index 0) et on ajoute user_nb à la fin

print(list2)
````

# Mission 4

# Mission 6

Ask the user for a string, convert it to a list of characters with the `list` function, then ask for a single character.

Remove the first occurrence of the character from the list, then display how many occurrences of the character remain in the list.

For instance if you enter `"Excellent work"` then the letter `"e"`, it will display:
```
There are still 2 copies of e in the list
```

````python
user_string3 = input("Specify a string / sentance : ")  # Input utilisateur d'une phrase
list_user_string3 = list(user_string3.lower())  # On crée une variable qui contient la phrase utilisateur sous forme de liste. On passe en minuscule la string pour éviter les problèmes de casse (entre "E" et "e" par exemple)
user_char = input("Specify a single character you want to count in your sentance : ")  # On demande une lettre à l'utilisateur
compteur = 0

list_user_string3.pop(list_user_string3.index(user_char)) # On retire (pop) la première occurence de la lettre spécifiée que l'on trouve grâce à .index()

for i in list_user_string3: # On peut maintenant compter le nombre de fois que la lettre spécifiée est présente dans la phrase saisie
  if i == user_char:
    compteur += 1 # Sans oublier d'incrémenter le compteur si la condition est vraie

# On peut aussi utiliser la méthode .count() des strings :
# comptage = list_user_string3.count("e")


print(f"There are still {compteur} copies of '{user_char}' in the list")
````

# Mission 7

Define three lists containing no duplicates. Then, indicate (however you like) the two that have the most elements in common (or all three if there's a tie).

For instance, with:
```
list_a = [1, 2, 4, 8, 16, 32]
list_b = [1, 2, 3, 5, 8, 13]
list_c = [2, 3, 5, 7, 11, 13]
```
It may display:
```
Lists B and C have the most common elements!
```

````python
list_a = [1, 2, 4, 4, 16, 32] # On crée nos 3 listes à l'intérieur desquelles il n'y a pas de doublons
list_b = [1, 2, 3, 5, 8, 13]
list_c = [2, 3, 5, 7, 11, 13]

compteur_ab = 0 # On crée 3 variables qui permettront de compter le nb d'éléments communs entre les listes A, B et C
compteur_ac = 0
compteur_bc = 0

for a in list_a:  # La première boucle compare les éléments de la liste A avec les listes B et C
  for b in list_b:
    if a == b:
      compteur_ab += 1
  for c in list_c:
    if a == c:
      compteur_ac += 1

for bc in list_b: # La deuxième boucle compare les listes B et C. Chacune a déjà été comparée à A dans la première boucle)
  for cb in list_c:
    if bc == cb:
      compteur_bc += 1

most_common = max(compteur_ab, compteur_ac, compteur_bc)  # Création d'une variable qui stocke le nb du compteur le plus grand

if compteur_ab == compteur_ac or compteur_ab == compteur_bc or compteur_ac == compteur_bc:  # On vérifie que les listes ne partagent pas toutes 3 le même nb d'éléments en commun
  print(f"The three lists share the most common elements : {most_common}")
elif most_common == compteur_ab:  # Si ce n'est pas le cas, suivant le compteur correspondant au maximum des 3, on affiche un message différent
  print(f"Lists A and B have the most common elements : {compteur_ab}")
elif most_common == compteur_ac:
  print(f"Lists A and C have the most common elements : {compteur_ac}")
elif most_common == compteur_bc:
  print(f"Lists B and C have the most common elements : {compteur_bc}")
````
