# 1. Calcul d'un prix

Ecrivez un programme faisant saisir à l’utilisateur un prix unitaire HT, un taux de TVA et un nombre d’articles, et qui calcule le montant TTC de l’achat.<br>
Affichez le montant TTC.

```python
price_vat_excl = float(input("Saisissez le prix HT unitaire d'un article"))
vat_rate = float(input("Quel est le taux de TVA ? Par ex : 5.5 pour 5,5%"))
article_nb = float(input("Combien d'articles sont concernés ?"))

price_vat_incl = article_nb * (price_vat_excl * (1 + vat_rate / 100))

print(f"Le prix total TTC est de {round(price_vat_incl,2)}€")
```

# 2. Conversion de durées (1)

Ecrivez un programme demandant une durée en secondes à l’utilisateur, et qui la convertit en heures, minutes, secondes.<br>
Affichez le résultat.<br>
Par exemple 12334 deviendra 3 heures, 25 minutes et 34 secondes.

```python
seconds = int(input("Indiquez une durée en secondes"))

hours = seconds // 3600
seconds -= 3600 * hours
minutes = seconds // 60
seconds -= 60 * minutes

print(f"Cela correspond à {hours} heures, {minutes} minutes et {seconds} secondes")
```

# 3. Conversion de durées (2)

Ecrivez un programme demandant une durée en heures, minutes, secondes à l’utilisateur et qui la convertit en secondes.<br>
Affichez le résultat.<br>
Par exemple 3 heures, 25 minutes et 34 secondes deviendra 12334 secondes.

```python
hours = int(input("Vous allez renseigner une durée en heures, minutes et secondes.\nVeuillez commencer par renseigner les heures :"))
minutes = int(input("Veuillez maintenant renseigner les minutes :"))
seconds = int(input("Pour finir, renseignez le nombre de secondes :"))

seconds_final = (3600 * hours) + (60 * minutes) + seconds

print(f"La durée renseignée correspond à {seconds_final} secondes")
```
