2. What does `i` correspond to?

Answer:

Answer:  
Because the first variable named in the "for" loop is val which takes the index values.
No matter the "for" variables name, the first one will return the index / counter and the second will return the values

2. What does `range(4)` mean?  
Why are only the first 4 elements of `test_list` displayed?

Answer:  
The range(4) means we are using a "for" loop with i from 0 to 4 excluded.
Only the first 4 elements are displayed because the range function is basically cycling like this : 0, 1, 2 and 3 which represents a total of 4 values.

```python
for i in range(2, 8, 2):
  print(test_list[i])
```

3. What does `range(2,8,2)` mean ?  
Why are only the elements 'name', 'hi' and 'name' displayed?

Answer:  
It means to go from the second element up to the eight one excluded with a step of 2.
Basically it goes like : 2, 4 and 6. 'name', 'hi' and 'name' have respectively indexes 2, 4 and 6 and this is why only those are displayed.

```python
test_list = ['Hi', 'my', 'name', 'is', "hi", "my","name", "is", "chickychicky"]

test_list = " ".join(test_list)

test_list
```

1. What can the `.join()` method do?  
What is the type of `test_list` before using `.join()`?  
What is the type of `test_list` after using `.join()`?

Answer:  
The .join() method joins all the items in an iterable (here, the test_list list) into one string using the specified parameter as the separator (here " ").

Before using this method, test_list was a list, after the operation it is a string object.

```python
christmas = "Noël c'est bien, mais attention à ne pas manger trop de chocolat. On le sait, mais on le fais à chaque fois. C'est la vie"

christmas.split()
```

2. What can the `.split()` method do?  
What is the type of `christmas` before using `.split()`?  
What is the type of `christmas` after using `.split()`?

Answer:  
The .split() method is the opposite of the .join() method : it splits the specified string into a list. The default separator is " " but you can specify one to your liking.

Before the split, christmas was a string and it became a list afterwards.

```python
christmas = "Noël c'est bien, mais attention à ne pas manger trop de chocolat. On le sait, mais on le fait à chaque fois. C'est la vie"


christmas.split("à")
```

3. Why does the `.split()` method split the `string` into 3 elements?

Answer:  
Because this time we didn't use the default separator (i.e. a blankspace) but instead we set "à" as the separator.

By doing so, we split christmas string into three part because the original string contained 3 times the "à" string.

Note that the separator used does not appear in the list formed by the .split() method.
